```python
import pandas as pd
```

This cell loads the cleaned climate datasets (temperature, CO2, and sea level) from the project folder.

```python
temp = pd.read_csv("../data/cleaned/temp_clean.csv")
co2 = pd.read_csv("../data/cleaned/co2_clean.csv")
sea = pd.read_csv("../data/cleaned/sea_clean.csv")
```

This cell checks the year ranges in each dataset so we know whether they overlap for a fair comparison.

```python
print(temp['Year'].min(), temp['Year'].max())
print(co2['Year'].min(), co2['Year'].max())
print(sea['Year'].min(), sea['Year'].max())
```

```python
common_years = set(temp['Year']) & set(co2['Year']) & set(sea['Year'])

temp = temp[temp['Year'].isin(common_years)]
co2 = co2[co2['Year'].isin(common_years)]
sea = sea[sea['Year'].isin(common_years)]
```

## Merging of the three datasets

```python
df = pd.merge(temp, co2, on='Year', how='inner')
df = pd.merge(df, sea, on='Year', how='inner')
df
```

```python
df = df.sort_values('Year')
df = df.dropna()
df.reset_index(drop=True, inplace=True)
```
