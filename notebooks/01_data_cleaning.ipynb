```python
import pandas as pd
```

```python
temp = pd.read_csv("../data/raw/global_temp.csv")
co2 = pd.read_csv("../data/raw/co2.csv")
sea = pd.read_csv("../data/raw/sea_level.csv")
```

Temperature Dataset Cleaning

```python
temp.head()
temp.info()
temp.isnull().sum()
```

```python
temp['Year'] = pd.to_datetime(temp['dt'], errors='coerce').dt.year
temp = temp.dropna(subset=['AverageTemperature', 'Year'])
temp_clean = temp.groupby('Year')['AverageTemperature'].mean().reset_index()
temp_clean.rename(columns={'AverageTemperature': 'Temperature'}, inplace=True)
```

```python
temp_clean.head()
```

```python
temp_clean.to_csv("../data/cleaned/temp_clean.csv", index=False)
```

CO2 Dataset Cleaning

```python
co2.head()
co2.columns
```

```python
co2 = co2[['year', 'value']]
co2.rename(columns={
    'year': 'Year',
    'value': 'CO2'
}, inplace=True)
```

```python
co2 = co2.dropna()
co2_clean = co2.groupby('Year')['CO2'].mean().reset_index()
```

```python
co2_clean.to_csv("../data/cleaned/co2_clean.csv", index=False)
```

SEA LEVEL DATASET CLEANING

```python
sea.head()
sea.columns
```

```python
sea.rename(columns={
    'year': 'Year',
    'mmfrom1993-2008average': 'SeaLevel'
}, inplace=True)
```

```python
sea = sea.dropna(subset=['Year', 'SeaLevel'])
sea['Year'] = sea['Year'].astype(int)
sea['SeaLevel'] = sea['SeaLevel'].astype(float)
```

```python
sea_clean = sea.groupby('Year')['SeaLevel'].mean().reset_index()
sea_clean = sea_clean.sort_values('Year')
```

```python
sea_clean.head()
```

```python
sea_clean.to_csv("../data/cleaned/sea_clean.csv", index=False)
```

```python
sea_clean['SeaLevel_MA'] = sea_clean['SeaLevel'].rolling(5).mean()
```
