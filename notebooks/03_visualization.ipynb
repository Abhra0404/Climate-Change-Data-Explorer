```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/cleaned/final_climate_data.csv")

plt.style.use('seaborn-v0_8')
```

## MULTI-TREND CLIMATE GRAPH

```python
df_norm = (df - df.min()) / (df.max() - df.min())

plt.figure(figsize=(12,6))
plt.plot(df['Year'], df_norm['Temperature'], label='Temperature')
plt.plot(df['Year'], df_norm['CO2'], label='CO2')
plt.plot(df['Year'], df_norm['SeaLevel'], label='Sea Level')

plt.title("Normalized Climate Trends", fontsize=14)
plt.legend()
plt.show()
```

## CORRELATION HEATMAP

```python
plt.figure(figsize=(8,6))

sns.heatmap(df[['Temperature','CO2','SeaLevel']].corr(),
            annot=True,
            cmap='coolwarm',
            fmt=".2f")

plt.title("Correlation Between Climate Variables")
plt.show()
```

## CO₂ vs TEMPERATURE (CAUSE ANALYSIS)

```python
plt.figure(figsize=(8,6))

sns.regplot(x='CO2', y='Temperature', data=df, scatter_kws={'alpha':0.6})

plt.title("CO2 vs Temperature Relationship")
plt.xlabel("CO2 Emissions")
plt.ylabel("Temperature")

plt.show()
```

## TEMPERATURE vs SEA LEVEL (IMPACT GRAPH)

```python
plt.figure(figsize=(8,6))

sns.regplot(x='Temperature', y='SeaLevel', data=df, scatter_kws={'alpha':0.6})

plt.title("Temperature vs Sea Level Rise")
plt.xlabel("Temperature")
plt.ylabel("Sea Level")

plt.show()
```

## ROLLING AVERAGE TREND (SMOOTH GRAPHS)

```python
df['Temp_MA'] = df['Temperature'].rolling(5).mean()

plt.figure(figsize=(12,6))
plt.plot(df['Year'], df['Temperature'], alpha=0.5, label='Raw')
plt.plot(df['Year'], df['Temp_MA'], linewidth=2, label='Smoothed')

plt.legend()
plt.title("Temperature Trend with Moving Average")
plt.show()
```
