```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("../data/cleaned/final_climate_data.csv")
```

```python
years = df['Year']
temps = df['Temperature']

coeffs = np.polyfit(years, temps, 1)  # degree 1 = linear
trend = np.poly1d(coeffs)
```

```python
future_years = np.arange(df['Year'].max()+1, df['Year'].max()+20)

future_temps = trend(future_years)
```

```python
plt.figure(figsize=(12,6))

plt.plot(years, temps, label="Actual", linewidth=2)
plt.plot(future_years, future_temps, linestyle='--', label="Predicted", linewidth=2)

plt.title("Temperature Prediction")
plt.xlabel("Year")
plt.ylabel("Temperature")
plt.legend()
plt.grid(True)

plt.show()
```

```python
sea = df['SeaLevel']

coeffs_sea = np.polyfit(years, sea, 1)
trend_sea = np.poly1d(coeffs_sea)

future_sea = trend_sea(future_years)
```

```python
plt.figure(figsize=(12,6))

plt.plot(years, sea, label="Actual Sea Level")
plt.plot(future_years, future_sea, linestyle='--', label="Predicted Sea Level")

plt.legend()
plt.title("Sea Level Prediction")
plt.show()
```

```python
year = 2035
pred_temp = trend(year)

print(f"Predicted temperature in {year}: {pred_temp:.2f}")
```

```python
from sklearn.metrics import r2_score

predicted_current = trend(years)
print("R2 Score:", r2_score(temps, predicted_current))
```

```python
coeffs2 = np.polyfit(years, temps, 2)
trend2 = np.poly1d(coeffs2)

plt.plot(years, temps, label="Actual")
plt.plot(years, trend2(years), label="Polynomial Fit")

plt.legend()
plt.show()
```
